```python
import numpy as np
import cv2
import os

def binarize_mnist_image(image_path, threshold=50):
    """读取 MNIST 图片并二值化处理"""
    # 检查文件是否存在
    if not os.path.exists(image_path):
        print(f"File not found: {image_path}")
        return None
    
    img = cv2.imread(image_path)
    if img is None:
        print(f"Failed to load image: {image_path}")
        return None
    
    # 先转灰度，再做二值化
    img_grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    _, img_binary = cv2.threshold(img_grey, threshold, 255, cv2.THRESH_BINARY)
    
    print(f"Shape: {img_binary.shape}")
    return img_binary

if __name__ == '__main__':
    result = binarize_mnist_image("MNIST_IMG_TOP9/test/5/1022.jpg")
    if result is not None:
        np.savetxt("img_data.txt", result, fmt="%d")
        print("Saved to img_data.txt")
```
