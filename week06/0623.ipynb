# MySQL 基础语法

## 1. 数据库操作

### 创建数据库
```sql
CREATE DATABASE database_name;
```

### 删除数据库
```sql
DROP DATABASE database_name;
```

### 选择数据库
```sql
USE database_name;
```

### 查看所有数据库
```sql
SHOW DATABASES;
```

---

## 2. 表操作

### 创建表
```sql
CREATE TABLE table_name (
    column1 datatype constraints,
    column2 datatype constraints,
    ...
);
```

**示例：**
```sql
CREATE TABLE students (
    id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(50) NOT NULL,
    age INT,
    gender CHAR(1),
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

### 删除表
```sql
DROP TABLE table_name;
```

### 修改表（添加列）
```sql
ALTER TABLE table_name ADD column_name datatype;
```

### 查看表结构
```sql
DESC table_name;
```

---

## 3. 数据操作（CRUD）

### 插入数据（INSERT）
```sql
INSERT INTO table_name (column1, column2, ...) VALUES (value1, value2, ...);
```

**示例：**
```sql
INSERT INTO students (name, age, gender) VALUES ('张三', 20, 'M');
```

### 查询数据（SELECT）
```sql
SELECT column1, column2 FROM table_name WHERE condition;
```

**示例：**
```sql
SELECT * FROM students;
SELECT name, age FROM students WHERE age > 18;
```

### 更新数据（UPDATE）
```sql
UPDATE table_name SET column1 = value1 WHERE condition;
```

**示例：**
```sql
UPDATE students SET age = 21 WHERE name = '张三';
```

### 删除数据（DELETE）
```sql
DELETE FROM table_name WHERE condition;
```

**示例：**
```sql
DELETE FROM students WHERE id = 1;
```

---

## 4. 常用子句

### WHERE 条件过滤
```sql
SELECT * FROM students WHERE age >= 18 AND gender = 'M';
```

### ORDER BY 排序
```sql
SELECT * FROM students ORDER BY age DESC;  -- 降序
SELECT * FROM students ORDER BY age ASC;   -- 升序（默认）
```

### LIMIT 限制条数
```sql
SELECT * FROM students LIMIT 5;        -- 前5条
SELECT * FROM students LIMIT 5 OFFSET 2; -- 跳过2条，取5条
```

### GROUP BY 分组
```sql
SELECT gender, COUNT(*) FROM students GROUP BY gender;
```

### HAVING 分组后过滤
```sql
SELECT gender, COUNT(*) AS cnt FROM students GROUP BY gender HAVING cnt > 1;
```

---

## 5. 常用聚合函数

| 函数 | 说明 |
|------|------|
| `COUNT()` | 计数 |
| `SUM()` | 求和 |
| `AVG()` | 平均值 |
| `MAX()` | 最大值 |
| `MIN()` | 最小值 |

**示例：**
```sql
SELECT AVG(age) AS avg_age, MAX(age) AS max_age FROM students;
```

---

## 6. 多表查询

### 内连接（INNER JOIN）
```sql
SELECT a.name, b.score
FROM students a
INNER JOIN scores b ON a.id = b.student_id;
```

### 左连接（LEFT JOIN）
```sql
SELECT a.name, b.score
FROM students a
LEFT JOIN scores b ON a.id = b.student_id;
```

---

## 7. 索引

```sql
-- 创建索引
CREATE INDEX idx_name ON table_name (column_name);

-- 查看索引
SHOW INDEX FROM table_name;

-- 删除索引
DROP INDEX idx_name ON table_name;
```

# SQL 实战练习：学生成绩管理系统

本教程通过一个完整的**学生成绩管理**案例，演示 MySQL 核心 SQL 语法。包含建表、插入数据、单表查询、多表查询、集合运算等操作。

---

## 1. 建表（DDL）

### 创建学生信息表

包含学号、姓名、性别、出生年份、院系、家庭住址等字段。

```sql
CREATE TABLE Student (
    Id INT(10) AUTO_INCREMENT PRIMARY KEY COMMENT '学号',
    Name VARCHAR(20) NOT NULL COMMENT '姓名',
    Sex VARCHAR(4) COMMENT '性别',
    Birth YEAR COMMENT '出生年份',
    Department VARCHAR(20) NOT NULL COMMENT '院系',
    Address VARCHAR(50) COMMENT '家庭住址'
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4 COMMENT='学生信息表';
```

### 创建成绩表

包含编号、学号、课程名、分数等字段，通过 `StuId` 关联学生表。

```sql
CREATE TABLE Score (
    Id INT(10) AUTO_INCREMENT PRIMARY KEY COMMENT '编号',
    StuId INT(10) NOT NULL COMMENT '学号',
    CName VARCHAR(20) COMMENT '课程名',
    Grade INT(10) COMMENT '分数'
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4 COMMENT='成绩表';
```

## 3. 单表查询

### 3.1 查询所有学生信息

```sql
SELECT * FROM Student;
```

### 3.2 分页查询

从第 2 条记录开始查询 3 条数据（`LIMIT 偏移量, 条数`）。

```sql
SELECT * FROM Student
LIMIT 1, 3;
```

### 3.3 查询指定列并设置别名

使用 `AS` 关键字为字段设置中文别名，让结果更易读。

```sql
SELECT Id AS 学号, Name AS 姓名, Department AS 院系
FROM Student;
```

### 3.4 条件查询（IN）

查询院系为"计算机系"或"英语系"的学生。

```sql
SELECT * FROM Student
WHERE Department IN ('计算机系', '英语系');
```

### 3.5 范围查询（BETWEEN）

查询出生年份在 2004~2008 年之间的学生。

```sql
SELECT * FROM Student
WHERE Birth BETWEEN 2004 AND 2008;
```

### 3.6 分组统计（GROUP BY）

按院系统计各院系的学生人数。

```sql
SELECT Department AS 院系, COUNT(*) AS 人数
FROM Student
GROUP BY Department;
```

### 3.7 聚合函数 — 最高分

查询各科目的最高分。

```sql
SELECT CName AS 科目, MAX(Grade) AS 最高分
FROM Score
GROUP BY CName;
```

### 3.8 聚合函数 — 平均分

查询各科目的平均成绩，保留两位小数。

```sql
SELECT
    CName AS 科目,
    ROUND(AVG(Grade), 2) AS 平均成绩
FROM Score
GROUP BY CName;
```

## 4. 多表查询（JOIN）

### 4.1 内连接（INNER JOIN）

查询"李四"的考试成绩，需关联 `Student` 和 `Score` 两张表。

```sql
SELECT sc.CName AS 考试科目, sc.Grade AS 考试成绩
FROM Score sc
INNER JOIN Student s ON sc.StuId = s.Id
WHERE s.Name = '李四';
```

### 4.2 左连接（LEFT JOIN）

查询所有学生及其考试成绩。**未参加考试的学生也会显示**（成绩字段为 NULL）。

```sql
SELECT s.*, sc.CName, sc.Grade
FROM Student s
LEFT JOIN Score sc ON s.Id = sc.StuId;
```

### 4.3 多表分组统计

查询每位学生的总成绩。

```sql
SELECT
    s.Id AS 学号,
    s.Name AS 姓名,
    SUM(sc.Grade) AS 总成绩
FROM Student s
INNER JOIN Score sc ON s.Id = sc.StuId
GROUP BY s.Id, s.Name;
```

### 4.4 多表条件过滤

查询计算机成绩低于 95 分的学生信息。

```sql
SELECT s.*
FROM Student s
INNER JOIN Score sc ON s.Id = sc.StuId
WHERE sc.CName = '计算机' AND sc.Grade < 95;
```

### 4.5 多表查询 + 排序

查询计算机成绩，按分数从高到低排序。

```sql
SELECT
    s.Name AS 姓名,
    sc.Grade AS 计算机成绩
FROM Score sc
INNER JOIN Student s ON sc.StuId = s.Id
WHERE sc.CName = '计算机'
ORDER BY sc.Grade DESC;
```

### 4.6 综合查询

查询地址中含有"湖南"的学生，计算其年龄并显示成绩详情。

```sql
SELECT
    s.Name AS 姓名,
    (2026 - s.Birth) AS 年龄,
    s.Department AS 院系,
    sc.CName AS 考试科目,
    sc.Grade AS 成绩
FROM Student s
INNER JOIN Score sc ON s.Id = sc.StuId
WHERE s.Address LIKE '%湖南%';
```

### UNION 并集

查询所有在**学生表**或**成绩表**中出现过的学号，自动去重。

```sql
SELECT Id AS StuId FROM Student
UNION
SELECT StuId FROM Score;
```

## 行转列：统计每日胜负次数

### 问题描述

原始表 `results` 记录每天的胜负结果：

| 日期 (rstime) | 结果 (result) |
|---------------|---------------|
| 2005-5-9      | 胜            |
| 2005-5-9      | 胜            |
| 2005-5-9      | 负            |
| 2005-5-9      | 负            |
| 2005-5-10     | 胜            |
| 2005-5-10     | 负            |
| 2005-5-10     | 负            |

需要转换为**行转列**的汇总表：

| 日期 (rstime) | 胜 | 负 |
|---------------|----|----|
| 2005-5-9      | 2  | 2  |
| 2005-5-10     | 1  | 2  |

### 建表并插入数据

```sql
-- 创建结果表
CREATE TABLE results (
    rstime DATE,
    result VARCHAR(10)
);

-- 插入示例数据
INSERT INTO results VALUES
    ('2005-5-9', '胜'),
    ('2005-5-9', '胜'),
    ('2005-5-9', '负'),
    ('2005-5-9', '负'),
    ('2005-5-10', '胜'),
    ('2005-5-10', '负'),
    ('2005-5-10', '负');
```

### 解法一：条件聚合（推荐）

使用 `SUM(CASE WHEN ...)` 将行值转为列：

```sql
SELECT
    rstime AS 日期,
    SUM(CASE WHEN result = '胜' THEN 1 ELSE 0 END) AS 胜,
    SUM(CASE WHEN result = '负' THEN 1 ELSE 0 END) AS 负
FROM results
GROUP BY rstime
ORDER BY rstime;
```

### 解法二：COUNT + IF

```sql
SELECT
    rstime AS 日期,
    COUNT(IF(result = '胜', 1, NULL)) AS 胜,
    COUNT(IF(result = '负', 1, NULL)) AS 负
FROM results
GROUP BY rstime
ORDER BY rstime;
```

### 解法三：COUNT + CASE

```sql
SELECT
    rstime AS 日期,
    COUNT(CASE WHEN result = '胜' THEN 1 END) AS 胜,
    COUNT(CASE WHEN result = '负' THEN 1 END) AS 负
FROM results
GROUP BY rstime
ORDER BY rstime;
```

### 执行结果

| 日期 (rstime) | 胜 | 负 |
|---------------|----|----|
| 2005-5-9      | 2  | 2  |
| 2005-5-10     | 1  | 2  |

> 💡 **核心思路：** 这种 **行转列（Pivot）** 场景，利用 `GROUP BY` 分组后，用 `CASE WHEN` 或 `IF` 做条件判断，结合聚合函数将行值"摊开"成多列。

## 分组统计：各地区合格/不合格人数及分数汇总

### 问题描述

原始表 `scores` 记录每个人的成绩和地区信息：

| ID | 姓名 | 状态 | 分数 | 地区 |
|----|------|------|------|------|
| 11 | wang | 不合格 | 60 | 重庆 |
| 12 | li   | 不合格 | 55 | 成都 |
| 13 | wang | 合格   | 80 | 重庆 |
| 14 | zhang| 合格   | 89 | 成都 |

需要按地区统计：

| 地区 | 合格的人数 | 不合格的人数 | 平均分数 | 总分 |
|------|-----------|-------------|----------|------|
| 成都 | 1         | 1           | 72.0     | 144  |
| 重庆 | 1         | 1           | 70.0     | 140  |

### 建表并插入数据

```sql
CREATE TABLE scores (
    ID INT,
    姓名 VARCHAR(20),
    状态 VARCHAR(10),
    分数 INT,
    地区 VARCHAR(10)
);

INSERT INTO scores VALUES
    (11, 'wang',  '不合格', 60, '重庆'),
    (12, 'li',    '不合格', 55, '成都'),
    (13, 'wang',  '合格',   80, '重庆'),
    (14, 'zhang', '合格',   89, '成都');
```

### 解法：分组 + 条件聚合

```sql
SELECT
    地区,
    SUM(CASE WHEN 状态 = '合格' THEN 1 ELSE 0 END) AS 合格的人数,
    SUM(CASE WHEN 状态 = '不合格' THEN 1 ELSE 0 END) AS 不合格的人数,
    ROUND(AVG(分数), 1) AS 平均分数,
    SUM(分数) AS 总分
FROM scores
GROUP BY 地区
ORDER BY 地区;
```

### 执行结果

| 地区 | 合格的人数 | 不合格的人数 | 平均分数 | 总分 |
|------|-----------|-------------|----------|------|
| 成都 | 1         | 1           | 72.0     | 144  |
| 重庆 | 1         | 1           | 70.0     | 140  |

> 💡 **核心思路：** `GROUP BY 地区` 分组后，用 `CASE WHEN` 统计各状态的计数，配合 `AVG()` 和 `SUM()` 聚合函数计算分数指标。
