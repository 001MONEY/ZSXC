Ollama本地大模型基础操作说明

- Windows端启动方法：下载安装完成后，在桌面找到羊驼图标应用，点击即可启动Ollama程序。Windows和macOS的操作逻辑略有区别。
- 初始配置建议：进入程序设置后，首先修改模型存储路径，默认存放在C盘，需要修改到其他磁盘；其次根据显存大小调整上下文长度，显存越大可以设置越长的上下文长度。
- 核心命令用法：使用`Ollama list`命令可以查看本地已经下载好的所有大模型；使用`Ollama ps`命令可以查看当前正在运行的模型，以及模型运行时CPU、GPU的占用比例；执行`ollama run 模型名称`命令，如果模型本地已存在会直接运行模型，如果不存在会自动下载后运行；输入`/bye`即可退出当前运行的模型。
- 模型下载规则：打开Ollama官网的Models页面，可以根据显存大小（8G/16G/32G/64G）选择对应模型，推荐新手使用qwen3.5 4B参数模型，4B参数模型可以在普通消费级显卡上正常运行。

EasyDataSet大语言模型数据集工具介绍与安装

- 工具核心功能：EasyDataSet是专门用于创建大语言模型微调数据集的工具，支持单轮问答、多轮问答、图片问答等多种数据集类型，支持PDF、Markdown、TXT等多种文档格式解析。
- 不同系统安装方式：打开工具的GitHub页面，可以根据操作系统选择安装包：Linux对应Linux安装包，老款Intel芯片Mac对应Intel安装包，新款M系列芯片Mac对应ARM安装包，Windows直接下载exe安装包即可，安装包大小仅200多兆，下载速度很快。
- 工具使用前置要求：安装后不需要注册登录，不登录也可以正常使用所有功能，注册登录反而增加操作步骤。

EasyDataSet模型配置方法

- 模型选择建议：创建项目后需要配置生成用的大模型，推荐选择阿里云百炼的qwen3.8 flash模型，该模型有免费额度，速度快效果好；也可以直接使用本地Ollama部署的模型，不需要消耗token额度。
- 阿里云百炼APIKey申请流程：搜索进入阿里云百炼大模型平台，登录账号后进入模型广场，找到APIKey入口点击创建，创建后需要立即复制或下载保存，关闭页面后无法再次查看完整APIKey。配置完成后刷新模型列表即可选择对应模型。
- 本地Ollama模型配置注意事项：配置本地Ollama模型时，必须保持Ollama程序处于运行状态，不能关闭；配置完成后点击刷新就能看到本地已经下载好的所有模型，选择对应模型保存后做测试，能正常回复即配置成功。

EasyDataSet单轮问答数据集制作流程

- 文档上传与切分：进入文件处理页面，上传任意格式文档，工具会自动将文档切分为多个文本块，本次演示上传峨眉山讲解词文档，自动切分为41个文本块。
- 文本块清洗与编辑：切分完成后可以选择自动清洗文本块，由大模型自动过滤不合格内容；清洗完成后点击编辑按钮，可以手动修改任意文本块内容，同时支持批量处理和单条修改，兼顾效率和灵活性。
- 生成与导出数据集：修改完成后可以批量或单条生成问答对，单个文本块默认生成4个问题，可以手动修改问题内容，确认后生成数据集，支持导出为JSON、CSV、ChatGPT、LLaMA Factory多种格式，满足不同微调框架的需求。本次导出为ChatGPT格式，添加了"你是一个旅游助手"的系统提示词，导出后格式符合标准要求。

EasyDataSet多轮对话数据集制作流程

- 任务参数配置：在项目设置的任务设置中，开启多轮对话数据集功能，需要设置系统提示词（定义AI助手的身份与规范）、对话场景、对话轮次（本次演示设置为5轮）、用户角色和助手角色。
- 生成注意事项：生成多轮对话数据集更消耗token额度，本次演示生成1组多轮对话就消耗了1%的可用token额度。
- 导出规则：多轮对话数据集默认导出为标准ChatGPT格式，不需要额外选择格式。

LLaMA Factory安装准备说明

- 环境要求：需要在带GPU的服务器或算力云节点上安装，公司普通服务器显卡驱动不支持，无法运行LLaMA Factory，推荐使用A10显卡的实例。
- 离线安装方式：官方教程要求用Git克隆仓库，为了方便学员，可以直接下载压缩包，上传到服务器解压使用，压缩包已经分享到学员群。
- 基础环境创建步骤：首先用conda创建独立环境，指定Python版本为3.10，激活环境后，将解压后的LLaMA Factory文件夹上传到服务器，进入文件夹后执行`pip install -e .`安装，一定要加上`-i`参数配置清华源，否则下载速度会很慢，作者自己安装的时候用默认源花了4个小时。

LavaFactory平台基础介绍

- 平台定位：LavaFactory是一款低代码大模型训练平台，降低了大模型微调的操作门槛。
- 推荐使用环境：推荐优先在Linux系统上使用，因为生产环境的服务器极少安装Windows系统，大模型训练也多在Linux服务器上进行。
- 登录设置说明：不同于ofdl平台使用平台统一分配的密码，LavaFactory支持用户自行设置登录密码。

LavaFactory本地环境搭建步骤

- conda环境创建要求：使用`conda create`创建环境，Python版本要求大于等于3.11，不支持3.10版本。
- 依赖安装要求：激活环境后，进入`llama_mine`文件夹，依次执行`pip install -e .`和`pip install -r`依赖文件。
- 换源注意事项：安装依赖建议配置清华源或中科大源，否则默认源下载速度极慢；同时需要注意，先安装PyTorch再换源安装其他依赖，避免换源后PyTorch被默认安装为CPU版本，导致训练时无法使用CUDA。
- 启动方式说明：平台支持命令行和图形界面两种启动方式，本次课程优先讲解图形界面操作。

LavaFactory前端界面参数配置说明

- 模型路径配置：支持从ModelScope或HuggingFace在线下载模型，也支持手动填入本地已下载模型的路径。
- LoRA与检查点配置：支持填入已训练好的LoRA检查点路径，支持从中断训练处继续训练。
- 量化配置：支持选择4比特或8比特量化，降低训练时的显存占用。
- 对话模板配置：训练对话模型必须选择对应的对话模板，保证训练数据格式符合模型要求。

自定义数据集添加流程

- 数据集存放位置：将自定义数据集放到`llama_factory/data`文件夹下。
- 数据集注册方法：在`data/dataset_info.json`文件中，复制已有数据集的配置项，修改名称和对应json文件名即可完成注册。
- 验证方法：注册完成后刷新界面，即可在数据集列表中找到新增的自定义数据集，还支持在线预览数据集内容。

模型下载与加载验证

- 模型选择原则：根据本地显卡显存大小选择对应参数规模的模型，本次课程演示选择了千问2.5 1.5B版本，适配11G显存的显卡。
- 从ModelScope下载模型：使用modelscope命令行工具下载模型到指定本地路径。
- 模型加载验证：在LavaFactory界面填入模型本地路径后，点击加载模型，加载完成后可以进行对话测试，验证模型是否可以正常使用。
- 训练前准备：验证完成后必须先卸载已加载的模型，释放显存，再开始训练，否则会出现显存不足的问题。

大模型微调训练操作

- 训练参数调整：主要调整LoRA相关参数，其他参数保持默认即可。
- 分布式训练支持：平台支持DeepSpeed分布式训练，单张显卡无法使用该功能。
- 训练启动与验证：点击开始训练后，若终端和界面正常输出损失函数值，则说明训练启动成功。本次演示使用2的batch size，训练3轮，在11G显存的2080Ti上成功启动训练，虽然2080Ti不支持bf16精度，但并未报错。

训练完成后模型导出

- 检查点选择：导出模型前必须选择训练完成的检查点，不选择则会导出原始未训练的模型。
- 分块与量化配置：可以自定义模型分块大小，支持导出时直接对模型进行量化，降低模型体积。导出路径支持用户自定义。
- 导出文件说明：导出完成后的模型文件夹包含模型权重、tokenizer词表、聊天模板等完整内容，可以直接上传到ModelScope或HuggingFace平台分享使用。
- 聊天模板说明：导出的模型中包含用Jinja语言编写的chat_template.jinja文件，该模板由LavaFactory根据训练时选择的对话模板自动生成，保存在tokenizer相关文件中，推理时会自动加载。

Olama使用常见问题解答

- Olama使用说明：Olama下载完成后不需要注册账号，直接使用命令行控制即可，注册账号浪费时间。
- 数据集路径问题：自定义数据集必须放到LavaFactory的data文件夹下，填写路径时需要填写从根目录开始的绝对路径。
- 文件分享问题：若直接下载速度慢，可以尝试通过微信群分享已经下载好的文件，若分享链接过期则需要重新分享。
