```python
# ========== 导入所需库 ==========
import torch
import torch.nn as nn          # PyTorch 神经网络模块
import torch.optim as optim    # 优化器（SGD、Adam 等）
import torchvision             # 计算机视觉工具库（数据集、模型等）
import torchvision.transforms as transforms  # 数据预处理/增强
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm          # 进度条

# 设置中文字体支持
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

# ========== 1. 加载 CIFAR-10 数据集 ==========
print("正在加载 CIFAR-10 数据集...")

# 训练集数据增强与归一化
# RandomCrop: 随机裁剪（填充4像素）增加多样性
# RandomHorizontalFlip: 随机水平翻转
# Normalize: 用 CIFAR-10 数据集的均值和标准差标准化
transform_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])

# 测试集只需 ToTensor 和归一化，不需要数据增强
transform_test = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])

# 加载训练集（从本地路径读取，不下载）
trainset = torchvision.datasets.CIFAR10(root='D:/Medicaldata', train=True, download=False, transform=transform_train)
# DataLoader: 批量加载、打乱、多进程加速
trainloader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=2)

# 加载测试集
testset = torchvision.datasets.CIFAR10(root='D:/Medicaldata', train=False, download=False, transform=transform_test)
testloader = torch.utils.data.DataLoader(testset, batch_size=100, shuffle=False, num_workers=2)

# CIFAR-10 的 10 个类别名称
classes = ('plane', 'car', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck')
print(f"训练集大小: {len(trainset)}, 测试集大小: {len(testset)}")

# ========== 2. 为 CIFAR-10 调整 VGG ==========
# CIFAR-10 图片为 32×32，比原始 VGG 的 224×224 小很多，需调整网络配置
# 主要改动：去掉最后两个 MaxPool，保留更大的特征图

def get_vgg_cifar_cfg(variant='VGG16'):
    """返回适合 CIFAR-10 的 VGG 配置（去掉最后两个 MaxPool，保留更大特征图）
       数字表示卷积层输出通道数，'M' 表示 MaxPool 下采样"""
    if variant == 'VGG11':
        return [64, 'M', 128, 'M', 256, 256, 'M', 512, 512, 'M', 512, 512]
    elif variant == 'VGG13':
        return [64, 64, 'M', 128, 128, 'M', 256, 256, 'M', 512, 512, 'M', 512, 512]
    elif variant == 'VGG16':
        return [64, 64, 'M', 128, 128, 'M', 256, 256, 256, 'M', 512, 512, 512, 'M', 512, 512, 512]
    elif variant == 'VGG19':
        return [64, 64, 'M', 128, 128, 'M', 256, 256, 256, 256, 'M', 512, 512, 512, 512, 'M', 512, 512, 512, 512]
    else:
        raise ValueError(f"Unknown variant: {variant}")

def make_vgg_cifar_layers(cfg, in_channels=3):
    """根据配置列表构建 VGG 的特征提取层（卷积 + BN + ReLU + 池化）"""
    layers = []
    for v in cfg:
        if v == 'M':
            # 'M' 表示最大池化层，尺寸减半
            layers.append(nn.MaxPool2d(kernel_size=2, stride=2))
        else:
            # 数字表示该层输出通道数，使用 3×3 小卷积核
            layers.append(nn.Conv2d(in_channels, v, kernel_size=3, padding=1))
            layers.append(nn.BatchNorm2d(v))     # 批归一化，加速收敛
            layers.append(nn.ReLU(inplace=True))  # ReLU 激活函数
            in_channels = v  # 下一层的输入通道数 = 当前层输出通道数
    return nn.Sequential(*layers)

class VGG_CIFAR(nn.Module):
    """专为 CIFAR-10 设计的 VGG 网络（输入 32×32 彩色图片）"""
    def __init__(self, variant='VGG16', num_classes=10):
        super().__init__()
        # 获取对应变体的配置，构建特征提取层
        cfg = get_vgg_cifar_cfg(variant)
        self.features = make_vgg_cifar_layers(cfg, in_channels=3)

        # 动态计算展平后的特征维度
        # 传入一个 dummy 张量做一次前向传播，得到特征图尺寸
        dummy = torch.zeros(1, 3, 32, 32)
        out = self.features(dummy)
        flat_dim = out.numel()  # 展平后的总元素数

        # 分类头（全连接层）
        # 比原始 VGG 更轻量（原始 VGG 用 4096→4096→1000）
        self.classifier = nn.Sequential(
            nn.Linear(flat_dim, 512),  # 第一层全连接
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),           # Dropout 防止过拟合
            nn.Linear(512, 512),       # 第二层全连接
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes),  # 输出层，10 个类别
        )

        # 初始化所有权重
        self._initialize_weights()

    def _initialize_weights(self):
        """自定义权重初始化：
           - Conv2d: Kaiming 正态初始化（适配 ReLU）
           - BatchNorm: weight=1, bias=0
           - Linear: 小范围正态分布"""
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                nn.init.normal_(m.weight, 0, 0.01)
                nn.init.constant_(m.bias, 0)

    def forward(self, x):
        """前向传播：特征提取 → 展平 → 分类"""
        x = self.features(x)          # 卷积层提取特征
        x = x.view(x.size(0), -1)     # 展平为二维 [batch, features]
        x = self.classifier(x)        # 全连接层分类
        return x

# ========== 3. 训练设置 ==========
# 优先使用 GPU（CUDA），否则回退到 CPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"使用设备: {device}")

# 实例化 VGG16 并移到 GPU/CPU
net = VGG_CIFAR('VGG16').to(device)
print(f"\n模型参数量: {sum(p.numel() for p in net.parameters())/1e6:.2f}M")
print(net)

# 损失函数：交叉熵损失（适合多分类）
criterion = nn.CrossEntropyLoss()
# 优化器：SGD + 动量 + 权重衰减（L2 正则化）
optimizer = optim.SGD(net.parameters(), lr=0.01, momentum=0.9, weight_decay=5e-4)
# 学习率调度：余弦退火，周期 T_max=100
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=100)

# ========== 4. 训练循环 ==========
num_epochs = 20  # 训练轮数（可增大到 100+ 获得更好效果）

train_losses = []   # 记录每个 epoch 的平均损失
test_accs = []      # 记录每个 epoch 的测试准确率
best_acc = 0.0      # 记录最佳准确率

for epoch in range(num_epochs):
    # === 训练阶段 ===
    net.train()                     # 切换到训练模式（启用 Dropout、BN 更新）
    running_loss = 0.0
    pbar = tqdm(trainloader, desc=f"Epoch {epoch+1}/{num_epochs}")

    for inputs, labels in pbar:
        inputs, labels = inputs.to(device), labels.to(device)  # 数据移到 GPU

        optimizer.zero_grad()   # 梯度清零
        outputs = net(inputs)   # 前向传播
        loss = criterion(outputs, labels)  # 计算损失
        loss.backward()         # 反向传播
        optimizer.step()        # 更新权重

        running_loss += loss.item()
        pbar.set_postfix({'loss': f'{loss.item():.4f}'})  # 更新进度条显示

    scheduler.step()  # 调整学习率

    # === 测试阶段 ===
    net.eval()                      # 切换到评估模式（关闭 Dropout、BN 固定）
    correct = total = 0
    with torch.no_grad():           # 不计算梯度，节省内存
        for inputs, labels in testloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = net(inputs)
            _, predicted = outputs.max(1)     # 取概率最大的类别
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()  # 统计正确数

    acc = 100.0 * correct / total
    avg_loss = running_loss / len(trainloader)
    train_losses.append(avg_loss)
    test_accs.append(acc)

    print(f"Epoch {epoch+1:2d} | Loss: {avg_loss:.4f} | Test Acc: {acc:.2f}% | LR: {scheduler.get_last_lr()[0]:.5f}")

    # 保存最佳模型
    if acc > best_acc:
        best_acc = acc
        torch.save(net.state_dict(), './vgg16_cifar10_best.pth')
        print(f"  → 保存最佳模型 (Acc: {acc:.2f}%)")

print(f"\n训练完成！最佳测试准确率: {best_acc:.2f}%")

# ========== 5. 可视化训练曲线 ==========
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# 左图：训练损失曲线
ax1.plot(range(1, num_epochs+1), train_losses, 'b-o', markersize=3)
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.set_title('训练损失曲线')
ax1.grid(True)

# 右图：测试准确率曲线
ax2.plot(range(1, num_epochs+1), test_accs, 'r-s', markersize=3)
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy (%)')
ax2.set_title('测试准确率曲线')
ax2.grid(True)

plt.suptitle('VGG16 在 CIFAR-10 上的训练结果', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# ========== 6. 展示部分测试结果 ==========
net.eval()
dataiter = iter(testloader)
images, labels = next(dataiter)                         # 取一批测试图片
images, labels = images.to(device), labels.to(device)
outputs = net(images)
_, predicted = outputs.max(1)                           # 模型预测结果

# 显示 10 张测试图片及其预测结果
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
axes = axes.flatten()
for i in range(10):
    # 将 Tensor 转换回 numpy 图片（反归一化）
    img = images[i].cpu().numpy().transpose((1, 2, 0))
    mean = np.array([0.4914, 0.4822, 0.4465])
    std = np.array([0.2023, 0.1994, 0.2010])
    img = img * std + mean
    img = np.clip(img, 0, 1)

    axes[i].imshow(img)
    axes[i].axis('off')
    # 绿色=预测正确，红色=预测错误
    color = 'green' if predicted[i] == labels[i] else 'red'
    axes[i].set_title(f"真:{classes[labels[i]]}\n预:{classes[predicted[i]]}", color=color, fontsize=9)

plt.suptitle('VGG16 CIFAR-10 预测结果 (绿色=正确, 红色=错误)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# ========== 用 ResNet18 训练 CIFAR-10 ==========

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# ========== 1. 加载 CIFAR-10 数据集 ==========
print("正在加载 CIFAR-10 数据集...")

# 训练集：数据增强（随机裁剪 + 水平翻转 + 归一化）
transform_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])

# 测试集：仅归一化
transform_test = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])

# 从本地目录加载（不下载）
trainset = torchvision.datasets.CIFAR10(root='D:/Medicaldata', train=True, download=False, transform=transform_train)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=2)

testset = torchvision.datasets.CIFAR10(root='D:/Medicaldata', train=False, download=False, transform=transform_test)
testloader = torch.utils.data.DataLoader(testset, batch_size=100, shuffle=False, num_workers=2)

classes = ('plane', 'car', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck')
print(f"训练集: {len(trainset)} 张, 测试集: {len(testset)} 张")

# ========== 2. 构建适应 CIFAR-10 的 ResNet18 ==========
# 原始 ResNet18 为 224×224 设计，CIFAR-10 为 32×32
# 修改点：
#   1. 将 7×7 conv(stride=2) → 3×3 conv(stride=1)，适配小图
#   2. 去掉首层 MaxPool，避免特征图过小
#   3. 全连接层输出改为 10 类

def get_resnet18_cifar(num_classes=10):
    """加载标准 ResNet18 并调整适配 CIFAR-10"""
    model = torchvision.models.resnet18(weights=None, num_classes=num_classes)

    # 修改第一个卷积层：7×7(stride=2) → 3×3(stride=1)，去掉 padding=3
    model.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)

    # 去掉随后的最大池化层（identity 即跳过）
    model.maxpool = nn.Identity()

    # 修改全连接层输出为 10 类（原为 1000）
    in_features = model.fc.in_features
    model.fc = nn.Linear(in_features, num_classes)

    return model

# ========== 3. 训练设置 ==========
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"使用设备: {device}")

net = get_resnet18_cifar(num_classes=10).to(device)
total_params = sum(p.numel() for p in net.parameters())
print(f"ResNet18 参数量: {total_params/1e6:.2f}M")
print(net)

# 损失函数 + 优化器（SGD + 动量 + 权重衰减）
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(net.parameters(), lr=0.1, momentum=0.9, weight_decay=5e-4)
# 余弦退火学习率调度
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=200)

# ========== 4. 训练循环 ==========
num_epochs = 20  # 可增大到 200 获得更好效果

train_losses, test_accs = [], []
best_acc = 0.0

for epoch in range(num_epochs):
    # === 训练 ===
    net.train()
    running_loss = 0.0
    pbar = tqdm(trainloader, desc=f"Epoch {epoch+1}/{num_epochs}")

    for inputs, labels in pbar:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = net(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        pbar.set_postfix({'loss': f'{loss.item():.4f}'})

    scheduler.step()

    # === 测试 ===
    net.eval()
    correct = total = 0
    with torch.no_grad():
        for inputs, labels in testloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = net(inputs)
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()

    acc = 100.0 * correct / total
    avg_loss = running_loss / len(trainloader)
    train_losses.append(avg_loss)
    test_accs.append(acc)

    print(f"Epoch {epoch+1:2d} | Loss: {avg_loss:.4f} | Acc: {acc:.2f}% | LR: {scheduler.get_last_lr()[0]:.5f}")

    if acc > best_acc:
        best_acc = acc
        torch.save(net.state_dict(), './resnet18_cifar10_best.pth')
        print(f"  → 保存最佳模型 (Acc: {acc:.2f}%)")

print(f"\n训练完成！最佳准确率: {best_acc:.2f}%")

# ========== 5. 可视化训练曲线 ==========
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(range(1, num_epochs+1), train_losses, 'b-o', markersize=3)
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.set_title('训练损失曲线')
ax1.grid(True)

ax2.plot(range(1, num_epochs+1), test_accs, 'r-s', markersize=3)
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy (%)')
ax2.set_title('测试准确率曲线')
ax2.grid(True)

plt.suptitle('ResNet18 在 CIFAR-10 上的训练结果', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# ========== 6. 展示预测结果 ==========
net.eval()
dataiter = iter(testloader)
images, labels = next(dataiter)
images, labels = images.to(device), labels.to(device)
outputs = net(images)
_, predicted = outputs.max(1)

fig, axes = plt.subplots(2, 5, figsize=(12, 5))
axes = axes.flatten()
for i in range(10):
    img = images[i].cpu().numpy().transpose((1, 2, 0))
    mean = np.array([0.4914, 0.4822, 0.4465])
    std = np.array([0.2023, 0.1994, 0.2010])
    img = img * std + mean
    img = np.clip(img, 0, 1)

    axes[i].imshow(img)
    axes[i].axis('off')
    color = 'green' if predicted[i] == labels[i] else 'red'
    axes[i].set_title(f"真:{classes[labels[i]]}\n预:{classes[predicted[i]]}", color=color, fontsize=9)

plt.suptitle('ResNet18 CIFAR-10 预测结果 (绿=正确, 红=错误)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# ========== 用 ResNet34 训练 Oxford-IIIT Pet 二分类（猫 vs 狗） ==========

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import Dataset, DataLoader
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
import os
from tqdm import tqdm

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# ========== 1. 自定义数据集 ==========
# list.txt 格式：Image CLASS_ID SPECIES BREED_ID
# SPECIES: 1=猫, 2=狗 → 映射为标签 0, 1

class OxfordPetDataset(Dataset):
    """Oxford-IIIT Pet 数据集（二分类：猫 vs 狗）"""
    def __init__(self, root_dir, split='trainval', transform=None):
        self.img_dir = os.path.join(root_dir, 'images')
        self.transform = transform

        # 读取 list.txt 并过滤出指定 split 的数据
        all_lines = open(os.path.join(root_dir, 'annotations', 'list.txt')).read().strip().splitlines()
        split_lines = open(os.path.join(root_dir, 'annotations', f'{split}.txt')).read().strip().splitlines()
        split_ids = set(line.split()[0] for line in split_lines if not line.startswith('#'))

        self.samples = []
        for line in all_lines:
            if line.startswith('#'):
                continue
            parts = line.split()
            img_name = parts[0]          # 例如 Abyssinian_100
            species = int(parts[2])      # 1=猫, 2=狗
            if img_name in split_ids:
                self.samples.append((img_name, 0 if species == 1 else 1))

        print(f"  {split}: 猫={sum(1 for _,l in self.samples if l==0)}, "
              f"狗={sum(1 for _,l in self.samples if l==1)}")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_name, label = self.samples[idx]
        img_path = os.path.join(self.img_dir, f'{img_name}.jpg')
        image = Image.open(img_path).convert('RGB')

        if self.transform:
            image = self.transform(image)

        return image, label


# ========== 2. 数据预处理 ==========
# 图片尺寸不一，统一缩放到 224×224（ResNet34 标准输入）

transform_train = transforms.Compose([
    transforms.Resize(256),
    transforms.RandomResizedCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225)),  # ImageNet 均值/标准差
])

transform_test = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225)),
])

print("加载 Oxford-IIIT Pet 数据集...")
trainset = OxfordPetDataset('D:/oxford-iiit-pet', split='trainval', transform=transform_train)
testset  = OxfordPetDataset('D:/oxford-iiit-pet', split='test', transform=transform_test)

# ★ 修复：Windows 上 num_workers=0 避免多进程卡死
trainloader = DataLoader(trainset, batch_size=64, shuffle=True, num_workers=0)
testloader  = DataLoader(testset, batch_size=64, shuffle=False, num_workers=0)

print(f"训练集: {len(trainset)} 张, 测试集: {len(testset)} 张")

# ========== 3. 构建 ResNet34 ==========
def get_resnet34_pet(num_classes=2):
    """加载预训练的 ResNet34，修改分类头为二分类"""
    model = torchvision.models.resnet34(weights='IMAGENET1K_V1')

    # 替换分类头：原 1000 类 → 2 类（猫/狗）
    in_features = model.fc.in_features
    model.fc = nn.Sequential(
        nn.Linear(in_features, 256),
        nn.ReLU(inplace=True),
        nn.Dropout(0.3),
        nn.Linear(256, num_classes),
    )
    return model

# ========== 4. 训练设置 ==========
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"使用设备: {device}")

net = get_resnet34_pet(num_classes=2).to(device)
total_params = sum(p.numel() for p in net.parameters())
trainable_params = sum(p.numel() for p in net.parameters() if p.requires_grad)
print(f"总参数量: {total_params/1e6:.2f}M, 可训练: {trainable_params/1e6:.2f}M")

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(net.parameters(), lr=0.001, momentum=0.9, weight_decay=1e-4)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=50)

# ========== 5. 训练循环 ==========
num_epochs = 10  # 预训练模型收敛快

train_losses, test_accs = [], []
best_acc = 0.0

for epoch in range(num_epochs):
    # 训练
    net.train()
    running_loss = 0.0
    pbar = tqdm(trainloader, desc=f"Epoch {epoch+1}/{num_epochs}")

    for inputs, labels in pbar:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = net(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        pbar.set_postfix({'loss': f'{loss.item():.4f}'})

    scheduler.step()

    # 测试
    net.eval()
    correct = total = 0
    with torch.no_grad():
        for inputs, labels in testloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = net(inputs)
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()

    acc = 100.0 * correct / total
    avg_loss = running_loss / len(trainloader)
    train_losses.append(avg_loss)
    test_accs.append(acc)

    print(f"Epoch {epoch+1:2d} | Loss: {avg_loss:.4f} | Acc: {acc:.2f}% | LR: {scheduler.get_last_lr()[0]:.5f}")

    if acc > best_acc:
        best_acc = acc
        torch.save(net.state_dict(), './resnet34_pet_best.pth')
        print(f"  → 保存最佳模型 (Acc: {acc:.2f}%)")

print(f"\n训练完成！最佳准确率: {best_acc:.2f}%")

# ========== 6. 可视化 ==========
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(range(1, num_epochs+1), train_losses, 'b-o', markersize=3)
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.set_title('训练损失曲线')
ax1.grid(True)

ax2.plot(range(1, num_epochs+1), test_accs, 'r-s', markersize=3)
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy (%)')
ax2.set_title('测试准确率曲线')
ax2.grid(True)

plt.suptitle('ResNet34 Oxford Pet 二分类结果', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# 展示预测结果
net.eval()
dataiter = iter(testloader)
images, labels = next(dataiter)
images, labels = images.to(device), labels.to(device)
outputs = net(images)
_, predicted = outputs.max(1)

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
axes = axes.flatten()
class_names = ['Cat', 'Dog']
for i in range(10):
    img = images[i].cpu().numpy().transpose((1, 2, 0))
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    img = img * std + mean
    img = np.clip(img, 0, 1)

    axes[i].imshow(img)
    axes[i].axis('off')
    color = 'green' if predicted[i] == labels[i] else 'red'
    axes[i].set_title(f"真:{class_names[labels[i]]}\n预:{class_names[predicted[i]]}", color=color, fontsize=10)

plt.suptitle('ResNet34 猫狗分类预测 (绿=正确, 红=错误)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```
